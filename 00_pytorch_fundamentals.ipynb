## reshape , stakc, squeeze, unsqueeze
* reshape - reshapes to defined shape
* view - return view of input tensor but same memory (diff pov)
* stack - combine on top of each other
* squeeze - remove 1 dim from tensor
* unsqueeze - add 1 dim to tensor
* permuate - view the input with dimenion permute

```python
x = torch.arange(1.,10.)
x.shape
```

```python
x
```

```python
x_reshape = x.reshape(1,9)
x_reshape, x_reshape.shape
```

```python
x_reshape = x.reshape(9,1)
x_reshape, x_reshape.shape
```

```python
x_reshape = x.reshape(3,3)
x_reshape, x_reshape.shape
```

```python
z = x.view(1,9)
z, z.shape
```

```python
#chanigng z will change x
z[:, 0] = 5
z,x
```

```python
#stack tensors on top
x_stack = torch.stack([x,x,x,x], dim=0)
x_stack
```

```python
x_stack = torch.stack([x,x,x,x], dim=1)
x_stack
```

```python
x_reshape = x.reshape(1,9)
```

```python
print(f"Previous tensor: {x_reshape}")
print(f"Previous shape: {x_reshape.shape}")
```

```python
# Remove extra single dimension from x_reshaped
x_squeeze = x_reshape.squeeze()
print(f"\nNew tensor: {x_squeeze}")
print(f"New shape: {x_squeeze.shape}")
```

```python
#unsqueeze - add single dimenion
print(f"Previous tensor: {x_squeeze}")
print(f"Previous shape: {x_squeeze.shape}")

## Add an extra dimension with unsqueeze
x_unsqueezed = x_squeeze.unsqueeze(dim=1)
print(f"\nNew tensor: {x_unsqueezed}")
print(f"New shape: {x_unsqueezed.shape}")
```

```python
#permuate - rearnges dim
# Create tensor with specific shape
x_original = torch.rand(size=(225, 224, 3))

# Permute the original tensor to rearrange the axis order
x_permuted = x_original.permute(2, 0, 1) # shifts axis 0->1, 1->2, 2->0

print(f"Previous shape: {x_original.shape}")
print(f"New shape: {x_permuted.shape}")
```

```python
x_original[0,0,0] = 7238
x_original[0,0,0], x_permuted[0,0,0]
```

### reproducability

to reduce randomness in nn and pytorhc - random seed

seed flavor the randomness

```python
torch.rand(3,3)
```

```python
rand_a = torch.rand(3,4)
rand_b = torch.rand(3,4)

print(rand_a, rand_b, rand_a==rand_b)
```

```python
RANDOM_SEED = 42

torch.manual_seed(RANDOM_SEED)
rand_c = torch.rand(3,4)

torch.manual_seed(RANDOM_SEED)
rand_d = torch.rand(3,4)

print(rand_c, rand_d, rand_c==rand_d)
```
